## Data Types

After ingestion, all monthly passenger columns are stored as `object`.

This is expected because Eurostat files contain special value markers such as `:` and, in some years, additional flags/annotations.

As this project follows an ELT approach, no datatype conversion is performed during ingestion. Data type standardization will be handled later in the transformation layer (dbt).
