## Data Types

After ingestion, all monthly cargo columns are stored as `object`.

This is expected because Eurostat files contain special value markers such as `:` and, in some years, additional flags/annotations.

As this project follows an ELT approach, no datatype conversion is performed during ingestion. Data type standardization will be handled later in the transformation layer (dbt).

```python
excel_file = pd.ExcelFile("/Users/hendrikalbrecht/Neuefische/Capstone-Project/data/NL_Cargo_2024.xlsx")
print(excel_file.sheet_names)
```

```python
df_Cargo_2024_raw = pd.read_excel(
    "/Users/hendrikalbrecht/Neuefische/Capstone-Project/data/NL_Cargo_2024.xlsx",
    sheet_name="Sheet 1",
    header=8
)
```

```python
df_Cargo_2024_raw = df_Cargo_2024_raw.iloc[1:].reset_index(drop=True)
```

```python
df_Cargo_2024_raw = df_Cargo_2024_raw.dropna(axis=1, how="all")
```
